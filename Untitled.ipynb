```python
import requests, zipfile, io
import pandas as pd

def get_ipeds_admissions(year: int) -> pd.DataFrame:
    url = f"https://nces.ed.gov/ipeds/datacenter/data/ADM{year}.zip"
    r = requests.get(url)
    z = zipfile.ZipFile(io.BytesIO(r.content))
    csv_name = [f for f in z.namelist() if f.endswith(".csv")][0]
    with z.open(csv_name) as f:
        df = pd.read_csv(f, encoding="latin1")
        df.to_csv("data.csv", index=False)
    return df

# Example: Admissions data for 2022
#admissions_df = get_ipeds_admissions(year=2023)
#print(admissions_df.head())
#print("Shape:", admissions_df.shape)
```

```python
def get_institutions(year: int) -> pd.DataFrame:
    url = f"https://nces.ed.gov/ipeds/datacenter/data/HD{year}.zip"
    r = requests.get(url)
    z = zipfile.ZipFile(io.BytesIO(r.content))
    csv_name = [f for f in z.namelist() if f.endswith(".csv")][0]
    with z.open(csv_name) as f:
        df = pd.read_csv(f)
        df.to_csv("data.csv", index=False)
    return df
```

```python
import pandas as pd

def calculate_sat_totals(df):
    """Add total SAT scores (25th, 50th, 75th percentiles) to the dataframe."""
    df = df.copy()
    df["SAT Total 25th"] = df["SAT Verbal 25th Percentile"] + df["SAT Math 25th Percentile"]
    df["SAT Total 50th"] = df["SAT Verbal 50th Percentile"] + df["SAT Math 50th Percentile"]
    df["SAT Total 75th"] = df["SAT Verbal 75th Percentile"] + df["SAT Math 75th Percentile"]
    return df

def classify_score(score, p25, p75):
    """
    Classify a score as 'Reach', 'Target', or 'Safety' based on 25th and 75th percentiles.
    """
    if score < p25:
        return "Reach"
    elif score > p75:
        return "Safety"
    else:
        return "Target"

def classify_schools(final_admissions_df, sat_score=None, act_score=None):
    """
    Classifies schools into safety, target, or reach based on SAT or ACT score.
    """
    df = final_admissions_df.copy()
    
    if sat_score is not None:
        df = calculate_sat_totals(df)
        df["Admission Likelihood"] = df.apply(
            lambda row: classify_score(sat_score, row["SAT Total 25th"], row["SAT Total 75th"]),
            axis=1
        )
    elif act_score is not None:
        df["Admission Likelihood"] = df.apply(
            lambda row: classify_score(act_score, row["ACT Composite 25th Percentile"], row["ACT Composite 75th Percentile"]),
            axis=1
        )
    else:
        raise ValueError("You must provide either a SAT or ACT score.")
    
    return df[[
        "Institution Name", "City", "State", "ZIP Code", "Admission Likelihood"
    ]].sort_values(by="Admission Likelihood", key=lambda x: x.map({"Reach": 0, "Target": 1, "Safety": 2}))
```
